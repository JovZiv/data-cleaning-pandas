```python
import pandas as pd

#Raquel: ['Year', 'Fatal Y/N'***, 'Age', 'Injury', 'original order', 'Time', 'Activity']

url = "C:/Users/Raquel Marques/Desktop/Ironhack/Shark_attack/GSAF5 - Sheet1-GSAF.csv"

shark_attack = pd.read_csv(url)

print("Shape:",shark_attack.shape)

print("Data types:",shark_attack.dtypes)

print("Columns:",shark_attack.columns)



shark_attack.columns = shark_attack.columns.map(lambda x: x.lower().strip().replace(" ", "_"))

print("Columns:",shark_attack.columns)

shark_attack = shark_attack.drop(columns=["unnamed:_21", "href_formula", "href", "pdf", "case_number.1","original order"])
```

Year filtering

```python
shark_attack = shark_attack[shark_attack['year'] >= 2010]

print("Shape:",shark_attack.shape)
```

Fatal Y/N

```python
print("fatal_y/n:",shark_attack["fatal_y/n"].value_counts())

shark_attack['fatal_y/n'] = (shark_attack['fatal_y/n'].astype('string').str.strip().str.lower()).replace({'y x 2':'y','f':'y','m':'n'})

print("fatal_y/n:",shark_attack["fatal_y/n"].value_counts())
```
