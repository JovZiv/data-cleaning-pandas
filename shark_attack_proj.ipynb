```python
# Count the recorded shark attaks for each state and show the 15 states with the most attacks
sharkflix_df["state"].value_counts().head(15)
```

```python
# Define the 6 locations we selected for the SharkFlix analysis
top_6_states = [
    "Florida",
    "Hawaii",
    "California",
    "New South Wales",
    "Western Australia",
    "Queensland"
]

# Create a new DataFrame containing only these 6 locations
sharkflix_df = sharkflix_df[
    sharkflix_df["state"].isin(top_6_states)
].copy()
```

```python
# Count the recorded shark attacks in each of the 6 selected locations
sharkflix_df["state"].value_counts()
```
